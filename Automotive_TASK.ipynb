Grupy nazwane daty: Przeanalizuj datę w formacie YYYY-MM-DD i wyciągnij rok, miesiąc oraz dzień do słownika używając nazwanych grup ((?P<name>...)).

```python
import re

text="""RAPORT HARMONOGRAMU WDROŻENIA SYSTEMU (2026)

W dniach od 2026-01-15 do 2026-02-28 przeprowadzono testy wydajnościowe nowej bazy danych. Pierwszy etap wdrożenia zaplanowano na 2026-03-01, natomiast końcowa migracja danych odbędzie się 2026-04-15.

Wykaz ważnych terminów umownych:
- Podpisanie umowy: 2025-11-20
- Zapłata I raty (termin): 2025-12-01 (zrealizowano: 2025-11-28)
- Zapłata II raty (termin): 2026-06-30
- Wygaśnięcie gwarancji: 2028-12-31

Awarie i zgłoszenia serwisowe:
- Zgłoszenie #102: 2026-01-05 (zamknięte: 2026-01-06)
- Zgłoszenie #108: 2026-02-12 (zamknięte: 2026-02-14)
- Zgłoszenie #115: 2026-05-01 (w trakcie realizacji)"""
pattern=r"(?P<year>\d{4})-(?P<month>\d{2})-(?P<day>\d{2})"
result=[m.groupdict() for m in re.finditer(pattern, text)]
print(result)
```

Wyciąganie cen: Znajdź w tekście wszystkie kwoty wraz z walutą PLN lub USD (np. 150 PLN, 99.99 USD).

Walidacja silnego hasła: Stwórz wzorzec walidujący hasło: minimum 8 znaków, co najmniej jedna wielka litera, jedna cyfra i jeden znak specjalny.

```python
import re

password="strongpassword123"
pattern=r'^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@$!%*?&])[A-Za-z\d@$!%*?&]{8,}$'
result=re.search(pattern, password)
if result:
    print("silne haslo")
else:
    print('slabe hasło')
```

Lookbehind (Kwoty bez waluty): Wyciągnij z tekstu same wartości liczbowe, ale tylko te, które bezpośrednio poprzedzone są znakiem $ (użyj positive lookbehind (?<=...)).

```python
import re

text="""PODSUMOWANIE ZAMÓWIEŃ ZE SKLEPU W USA (SPRZĘT I AKCESORIA)

1. Laptop Dell XPS 15: $1499.99 (cena sugerowana przed rabatem: $1650)
2. Bezprzewodowa mysz pionowa: $29.95
3. Ergonomiczna podkładka pod mysz: $ 9.99
4. Słuchawki z redukcją szumów: $199.00 (rabat lojalnościowy: -$20.00)
5. Przejściówka USB-C na HDMI: $12.50
6. Naklejki na klawiaturę: $0.99
7. Przedłużenie gwarancji na 2 lata: $ 120.00
8. Zestaw monitorów 4K (2 sztuki): $2350.50

Koszt dostawy ekspresowej: $15.00
Łączna kwota do zapłaty: $4216.92"""
pattern=r'(?<=\$)\s*\d+(?:\.\d{2})?'
result=re.findall(pattern,text)
print(result)
```

Negative Lookahead: Znajdź słowo "Python", ale tylko w przypadkach, gdy NIE następuje po nim cyfra 2 ani 3 (użyj negative lookahead (?!...)).

```python
import re

text="""TESTOWANIE WZORCA PYTHON:

1. Wersja obecna: Używamy dzisiaj Python 3.12 w naszej firmie. (brak dopasowania)
2. Wersja starsza: Stary kod napisano w Python 2.7. (brak dopasowania)
3. Zapis bez spacji: Zainstaluj Python3 lub Python2 na serwerze. (brak dopasowania)

4. Czyste słowo: Język Python jest świetny do nauki. (DOPASOWANIE!)
5. Przyszłość: Czy kiedyś powstanie Python 4? (DOPASOWANIE!)
6. Inna cyfra: Mam zainstalowany Python 312 w ścieżce, ale też Python 1.5. (Python 312: brak, Python 1.5: DOPASOWANIE!)
7. Słowa w zdaniu: Tworzymy skrypty w języku Python do automatyzacji zadań. (DOPASOWANIE!)
8. Słowa pochodne: Doświadczony Pythonista pisze Pythonic code. (Pythonista: zignorowane przez \b)"""
pattern=r'\bPython\b(?!\s*[23])'
result=re.findall(pattern,text)
print(result)
```

Część 3: Analiza logów i generowanie raportu (pathlib)
Zapisz znalezione błędy do pliku w formacie: NazwaPliku | NumerLinii | TreśćBłędu

```python
import os
from pathlib import Path

folder=Path("zadanie_3_logi")
empty_file=Path("zadanie_3_logi/raport_bledow.txt")
empty_file.touch(exist_ok=True)
for root, dirs, files in os.walk(folder):
    for file in files:
        if files == 'raport_bledow.txt':
            continue
        path = Path(root) / file
        context = path.read_text(encoding='utf-8').splitlines()

        for number, line in enumerate(context, start=1):
            if 'ERROR' in line:
                with open('zadanie_3_logi/raport_bledow.txt', 'a', encoding='utf-8') as f:
                    f.write(f"{file}|{number}|{line}\n")
```
